## 📖 Project Overview & Objective

Large Language Models (LLMs) frequently generate plausible-sounding text containing factual inaccuracies, ungrounded assertions, or hallucinated citations. The **Generative AI Source Citation Assistant** is an academic framework designed to bridge the gap between generative text synthesis and rigorous academic verification.

### Key Capabilities
1. **Answer Synthesis**: Generates structured answers to complex academic research questions.
2. **Atomic Claim Extraction**: Decomposes long-form answers into discrete, verifiable factual claims.
3. **Authoritative Evidence Retrieval**: Searches reliable sources (`.edu`, `.gov`, UNESCO, WHO, IEEE, Nature, etc.) using domain authority weighting.
4. **Natural Language Entailment Validation**: Cross-checks evidence snippets against claims to assign verification statuses (`Supported`, `Partially Supported`, `Insufficient Evidence`, `Unsupported`).
5. **In-Text Citation Injection**: Dynamically links claims to evidence via in-text markers (`[1]`, `[2]`).
6. **Multi-Standard Bibliography**: Formats reference lists in **APA 7th**, **MLA 9th**, **IEEE**, and **BibTeX**.
7. **Interactive Gradio UI**: Demonstrates the full verification process through an intuitive web application.

---

### 🔄 Five-Stage Grounding Pipeline
```text
User Question
      │
      ▼
Answer Synthesis (Gemini / OpenAI / Fallback Engine)
      │
      ▼
Claim Extraction (Decomposes answer into atomic assertions)
      │
      ▼
Source Retrieval (DuckDuckGo + Wikipedia with .edu/.gov authority scoring)
      │
      ▼
Source Validation (NLI entailment evaluation & reasoning)
      │
      ▼
Citation Generation (In-text [1], [2] injection + APA/MLA/IEEE references)
      │
      ▼
Gradio Interface (Interactive demonstration & metrics)
```

## SECTION 6 — End-to-End Backend Demonstration

Execute an end-to-end academic question through the 5-stage pipeline to inspect claim extraction, source retrieval, entailment verification, and citation formatting.

````python
from pipeline import CitationAssistantPipeline

# Initialize pipeline instance
citation_pipeline = CitationAssistantPipeline()

test_query = "What are the effects of artificial intelligence on education?"
print("=" * 80)
print(f"🔬 EXECUTING TEST QUERY: '{test_query}'")
print(f"⚙️ Provider: {citation_pipeline.llm.provider.upper()} | Model: {citation_pipeline.llm.model_name}")
print("=" * 80 + "\n")

# Run pipeline
result = citation_pipeline.run(test_query, citation_style="APA")

print("=" * 80)
print("📝 1. GROUNDED ANSWER WITH IN-TEXT CITATIONS:")
print("=" * 80)
print(result.citation_result.cited_answer)

print("\n" + "=" * 80)
print("🔍 2. CLAIM VERIFICATION BREAKDOWN:")
print("=" * 80)
for i, val in enumerate(result.validations, start=1):
    status_symbol = "🟢" if val.status.value == "Supported" else ("🟡" if val.status.value == "Partially Supported" else "⚪")
    print(f"Claim {i} [{val.claim_id}]: {val.claim_text}")
    print(f"  Status:     {status_symbol} {val.status.value}")
    print(f"  Confidence: {val.confidence_score * 100:.0f}%")
    print(f"  Reasoning:  {val.reasoning}")
    if val.supporting_sources:
        for s in val.supporting_sources:
            print(f"  -> Source: [{s.publisher}] {s.title} (Authority Score: {s.authority_score}/100)")
            print(f"     URL:    {s.url}")
    print("-" * 60)

print("\n" + "=" * 80)
print(f"📚 3. FORMATTED REFERENCES ({result.citation_style} 7th Edition):")
print("=" * 80)
for ref in result.citation_result.references:
    print(ref.formatted_text)

print("\n" + "=" * 80)
print("📊 4. VERIFICATION METRICS:")
print("=" * 80)
print(f"Total Claims Evaluated:    {result.metrics.total_claims}")
print(f"Fully Supported Claims:    {result.metrics.supported_claims}")
print(f"Partially Supported:       {result.metrics.partially_supported_claims}")
print(f"Grounding Rate:            {result.metrics.grounding_rate_pct}%")
print(f"Unique Sources Cited:      {result.metrics.unique_sources_cited}")
print(f"Processing Time:           {result.metrics.processing_time_seconds}s")
print("=" * 80)
````
